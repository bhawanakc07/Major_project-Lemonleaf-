```python
#load both trained data
from tensorflow.keras.models import load_model

leaf_model = load_model("leaf_detector_model.keras")
disease_model = load_model("lemon_disease_model.keras")
```

```python
class_names = [
    'Anthracnose',
    'Bacterial Blight',
    'Citrus Canker',
    'Curl Virus',
    'Deficiency Leaf',
    'Dry Leaf',
    'Healthy Leaf',
    'Sooty Mould',
    'Spider Mites'
]
```

```python
#Final intergrated prediction
import numpy as np
from tensorflow.keras.preprocessing import image

def predict_disease(img_path):

    # load image
    img = image.load_img(img_path, target_size=(224,224))
    img_array = image.img_to_array(img)

    # normalize
    img_array = img_array / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    # -----------------------
    # STEP 1 → MODEL 1
    # lemon leaf detection
    # -----------------------
    leaf_prediction = leaf_model.predict(img_array)[0][0]

    print("\nLeaf Model Score:", leaf_prediction)

    # adjust logic based on your model output
    # (check if your model uses 0=leaf or 1=leaf)
    if leaf_prediction >= 0.5:
        print("❌ Not a Lemon Leaf")
        return

    print("✔ Lemon Leaf Detected")

    # -----------------------
    # STEP 2 → MODEL 2
    # disease classification
    # -----------------------
    disease_prediction = disease_model.predict(img_array)[0]

    class_index = np.argmax(disease_prediction)
    confidence = np.max(disease_prediction)

    disease_name = class_names[class_index]

    print("Disease:", disease_name)
    print("Confidence:", confidence)
```
