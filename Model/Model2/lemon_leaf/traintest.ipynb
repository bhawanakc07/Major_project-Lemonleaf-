```python
import os
import shutil
import random

input_folder = "lemon_leaf"
output_folder = "lemon_leaf_dataset"

train_ratio = 0.8

train_path = os.path.join(output_folder, "train")
test_path = os.path.join(output_folder, "test")

for root, dirs, files in os.walk(input_folder):

    if len(files) == 0:
        continue

    # get relative path structure
    rel_path = os.path.relpath(root, input_folder)

    train_dir = os.path.join(train_path, rel_path)
    test_dir = os.path.join(test_path, rel_path)

    os.makedirs(train_dir, exist_ok=True)
    os.makedirs(test_dir, exist_ok=True)

    images = [os.path.join(root, f) for f in files]
    random.shuffle(images)

    split_index = int(len(images) * train_ratio)

    train_imgs = images[:split_index]
    test_imgs = images[split_index:]

    # copy train
    for img in train_imgs:
        shutil.copy(img, train_dir)

    # copy test
    for img in test_imgs:
        shutil.copy(img, test_dir)

print("✅ STRUCTURE PRESERVED TRAIN/TEST SPLIT DONE")
```
