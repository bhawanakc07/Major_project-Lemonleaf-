```python
import os
from PIL import Image

folder_path = "not_lemon_leaf"

total_images = 0
corrupted = 0

print("\n==============================")
print("📊 NON-LEAF DATASET REPORT")
print("==============================")

for class_name in os.listdir(folder_path):

    class_path = os.path.join(folder_path, class_name)

    if not os.path.isdir(class_path):
        continue

    count = 0
    sizes = {}

    for img_name in os.listdir(class_path):

        img_path = os.path.join(class_path, img_name)

        try:
            img = Image.open(img_path)
            img.load()

            size = img.size   # (width, height)

            count += 1
            total_images += 1

            if size not in sizes:
                sizes[size] = 0
            sizes[size] += 1

        except:
            corrupted += 1

    print("\n======================")
    print("📂 Class:", class_name)
    print("======================")
    print("🖼️ Images:", count)

    for k, v in sizes.items():
        print(f"📏 Pixel Size: {k} → {v} images")

print("\n======================")
print("📊 TOTAL IMAGES:", total_images)
print("❌ CORRUPTED IMAGES:", corrupted)
print("======================")
```

```python
import os
import hashlib
from PIL import Image

folder_path = "not_lemon_leaf_resized"

hashes = set()
total_before = 0
total_after = 0
corrupted = 0
duplicates = 0

print("\n====================================")
print("📊 NON-LEAF DATA CLEANING REPORT")
print("====================================")

for class_name in os.listdir(folder_path):

    class_path = os.path.join(folder_path, class_name)

    if not os.path.isdir(class_path):
        continue

    class_count_before = 0
    class_count_after = 0
    pixel_sizes = {}

    print("\n======================")
    print("📂 Class:", class_name)
    print("======================")

    for img_name in os.listdir(class_path):

        img_path = os.path.join(class_path, img_name)

        class_count_before += 1
        total_before += 1

        try:
            # open image
            img = Image.open(img_path).convert("RGB")
            img.load()

            # check pixel size
            size = img.size
            pixel_sizes[size] = pixel_sizes.get(size, 0) + 1

            # check duplicate using hash
            img_hash = hashlib.md5(img.tobytes()).hexdigest()

            if img_hash in hashes:
                duplicates += 1
                os.remove(img_path)
                continue

            hashes.add(img_hash)

            # save clean image (overwrite)
            img.save(img_path)

            class_count_after += 1
            total_after += 1

        except:
            corrupted += 1
            try:
                os.remove(img_path)
            except:
                pass

    print("🖼️ Remaining Images:", class_count_after)

    for k, v in pixel_sizes.items():
        print(f"📏 Pixel Size: {k} → {v} images")

print("\n====================================")
print("📊 FINAL SUMMARY")
print("====================================")
print("Total Before Cleaning:", total_before)
print("Total After Cleaning:", total_after)
print("❌ Corrupted Removed:", corrupted)
print("♻️ Duplicates Removed:", duplicates)
print("====================================")
```
